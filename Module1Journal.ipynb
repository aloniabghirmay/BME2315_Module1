## Disease Background: 
Prevalence & Incidence
Alzheimer's disease is the most common cause of dementia and is known to affect older adults. As of 2026, about 7.4 million Americans aged 65 and older currently have Alzheimer's dementia. The prevalence of Alzheimer's continues to grow with age, for example, 5.2% of adults between the age of 65-74 have Alzheimer's dementia, compared to the 35.8% of adults aged 85 and older. Incidence, is the number of new cases that develop over a certain time period, while prevalence is how many people currently live with the disease. While some studies show that dementia incidence rates have declined in recent years, the number of people living with Alzheimer's is expected to increase.
https://www.alz.org/alzheimers-dementia/facts-figures#prevalence 

Economic Burden
With Alzheimer's disease becoming of the most common neurodegenerative disorders amongst the elderly, the global cost of dementia is expected to reach $2 trillion by 2030. There's a major financial burden on patients and families of said patients due to Alzheimer's disease. I mentioned the global cost by 2030, but in 2026 alone, the health and long term care of people with Alzheimer's is set set to reach $818 billion. Medical and long-term care make up $222 billion of this bill. Additionally, theres an ongoing issue with unpaid caregiving, in 2025, 13 million caregivers were unpaid, despite 19 billion hours of care. The economic impact of Alzheimer's is much more than just direct medical treatment, and creates a large burden.
https://pubmed.ncbi.nlm.nih.gov/37972428/ 
https://schaeffer.usc.edu/research/dementia-cost-us-818-billion-2026-study/ 

Risk Factors (Genetic, Lifestyle)
The risk of Alzheimer's disease is influenced by many facotrs such as genetics, medical situation, environment, and lifestyle, but, theres no single known cause for Alzheimer's. One of the strongest risk factors is age, as I mentioned above, there's a clear difference in the percentage of adults with Alzheimer's as you get older. Genetics has a massive impact as well, carrying the APOE e4 allele tends to make you more likely to develop Alzheimer's, but it doesnt guarantee that a person will develop the disease. Studies have also shown that certain ethnic groups tend to develop dementia more often than others. There's many lifestyle factors that play a huge role such as, lack of physical activity, smoking, and alchohol consumption. Maintaining cardiovascular health, staying active, and just overall healthy habits can contribute to reducing the risk of dementia. As both genetics and lifestyle contribute to risk, the development of Alzheimer's is known as the result of multiple influences rather than one singular facotr.
https://www.cdc.gov/alzheimers-dementia/prevention/index.html
