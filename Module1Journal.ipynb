## Disease Background: 
Prevalence & Incidence
Alzheimer's disease is the most common cause of dementia and is known to affect older adults. As of 2026, about 7.4 million Americans aged 65 and older currently have Alzheimer's dementia. The prevalence of Alzheimer's continues to grow with age, for example, 5.2% of adults between the age of 65-74 have Alzheimer's dementia, compared to the 35.8% of adults aged 85 and older. Incidence, is the number of new cases that develop over a certain time period, while prevalence is how many people currently live with the disease. While some studies show that dementia incidence rates have declined in recent years, the number of people living with Alzheimer's is expected to increase.
https://www.alz.org/alzheimers-dementia/facts-figures#prevalence 

Economic Burden
With Alzheimer's disease becoming of the most common neurodegenerative disorders amongst the elderly, the global cost of dementia is expected to reach $2 trillion by 2030. There's a major financial burden on patients and families of said patients due to Alzheimer's disease. I mentioned the global cost by 2030, but in 2026 alone, the health and long term care of people with Alzheimer's is set set to reach $818 billion. Medical and long-term care make up $222 billion of this bill. Additionally, theres an ongoing issue with unpaid caregiving, in 2025, 13 million caregivers were unpaid, despite 19 billion hours of care. The economic impact of Alzheimer's is much more than just direct medical treatment, and creates a large burden.
https://pubmed.ncbi.nlm.nih.gov/37972428/

Risk Factors (Genetic, Lifestyle)
The risk of Alzheimer's disease is influenced by many facotrs such as genetics, medical situation, environment, and lifestyle, but, theres no single known cause for Alzheimer's. One of the strongest risk factors is age, as I mentioned above, there's a clear difference in the percentage of adults with Alzheimer's as you get older. Genetics has a massive impact as well, carrying the APOE e4 allele tends to make you more likely to develop Alzheimer's, but it doesnt guarantee that a person will develop the disease. Studies have also shown that certain ethnic groups tend to develop dementia more often than others. There's many lifestyle factors that play a huge role such as, lack of physical activity, smoking, and alchohol consumption. Maintaining cardiovascular health, staying active, and just overall healthy habits can contribute to reducing the risk of dementia. As both genetics and lifestyle contribute to risk, the development of Alzheimer's is known as the result of multiple influences rather than one singular facotr.
https://www.cdc.gov/alzheimers-dementia/prevention/index.html

Societal determinants
There's many social determinants that can influence someones risk for Alzheimer's disease. Quality of education, access to healthcare, income and economic stability are just some of the factors that play a major role towards Alzheimer's. Not having access to quality healthcare can reduce the opportunity of spotting early signs of Alzheimer's and management of your condition. Referring back to economic burden of Alzheimer's, income and economic stabliity plays a major role in a person's risk for Alzheimer's, not being able to pay for treatments and medical care. Not everything is covered by health insurance. Symptoms
The most common early symptom of Alzheimer's is difficulty to remember information. Alzheimer's begins with memory and thinking issues that become more noticeable as time passes. As the disease progesses, basic everyday activities become harder and things like making decisions and solving problems become harder. In later stages of Alzheimer's, people often have difficulty recognizing familiy members, communication, and everyday activities without help.

Diagnosis
Alzheimer's disease is diagnosed through combining a patient's medical history, cognitive testing, physical & neurological exams, lab tests, and brain imaging. Doctors will test the patient's memory, problem solving, and other thinking abilities to determine whether or not Alzheimer's seems to be present. Tests like the Mini Mental State Examination (MMSE) can help to measure cognitive impairment, but as youll see later in our research, it's not enough to diagnose Alzheimer's diseases. Biomarkers that track beta amyloid and pTAU also help provide evidence for Alzheimer related changed. No single test can give you an answer, this is why a comprehensive evalution process is needed to come up with a conclusion.\https://www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-alzheimers-disease-diagnosed

Standard of care treatments (& reimbursement)
As of right now, theres no cure for Alzheimer's disease, but there are treatments that aid in managing the sympotms or even slowing down the development of the disease. Galantamine, benzgalatamine, rivastigmine, and donepezil are cholinesterase inhibitors which are used during the mild to moderate stages of Alzheimer's and they regulate glutamate activity within the brain. These inhibitors prevent the breakdown of a brain chemical, acetylcholine, which is important for memory and thinking. Also, recently, the drugs lecanemab and donanemab, which target beta amyloid plaques in the brain, have just been FDA-approved. Medicare provides coverage for some treatments that are anti-amyloid when some other requirements are met, but out of pocket costs can still occur.

Disease Progession & Prognosis
Within Alzheimer's disease, symptoms become more sever over time because the damage is spreading throughout the brain. In the mild stage, patients tend to experience memory loss and difficulty completing everyday tasks. In the moderate stage, memory gets worse and the patient becomes more confused, there's difficulty recognizing people, communicating, and doing daily tasks independently. In severe Alzheimer's, the damage is now spread throughout the brain and the patient will lose the ability to communicate and fully relies on others for care. The rate of this progression varies amongst individuals. Alzheimer's is a fatal disease because the brain damage then eventually intereferes with functions necessary for living.
https://www.nia.nih.gov/health/alzheimers-and-dementia/alzheimers-disease-fact-sheet
https://www.nia.nih.gov/health/alzheimers-and-dementia/alzheimers-disease-fact-sheet#stages-of-alzheimer-s-disease

Continuum of Care providers
People with Alzheimer's disease require aid from a variety of health care professionals, especially as the disease is progressing. At first, what's needed is a primary care physician who will evalute memory and the patient's health. But other specialists like nuerologists and geriatricians will diagnose and manage dementia that the patient suffers. Other health workers get involved as well, nurses, social workers, and therapists will help based off of the patient's needs. As the Alzheimer's gets worse, caregivers and family will play large roles in medication and daily activities. 
https://www.nia.nih.gov/health/alzheimers-treatment/next-steps-after-alzheimers-diagnosis

Biological Mechanisms (anatomy, organ physiology, cell & molecular physiology)
Alzheimer's disease causes damage to the brain and neurons. In the early stages, the damage is more in the hippocampus and entorhinal cortex, both of which are important for learning and memory. Beta amyloid and pTAU play a huge role in Alzheimer's and are the two major molecular features. Beta amyloid plaques outside of neurons and pTAU tangles within the neurons.  Usually, pTAU would stabilize the microtubules that transport material through neurons, but changes to it will cause it to detach and form tangles. This disrupts neuron communicatoin and normal cell functions, and eventually results in neurons dying and the brain shrinking. This is relevant to our project because we're analyzing pTAU and trying to link its relation to MMSE cognitive scores.

Clinical Trials/next-gen therapies

We mentioned that recently, lecanemab and donanemab have been approved, but now researchers are determining how useful it can be, and its utilization in other stages of Alzheimer's or if it can be combined with other treatments. As well, more research is going on that doesn't just primarily focus on beta amyloid, but also tau proteins, metabolism, the immune system, and neuronal function. The National Institute of Health, of Aging, and other Alzheimer's research programs are conducting hundreds of students and trying to dive deeper into the disease and about what can be done.

### i. Import Important Libraries, the Patient File, and the Data-Set

```python
# imports the Patient file and other libraries
from main_patients import *
import csv 
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
import math as math
from sklearn.linear_model import LinearRegression
```

### ii. Instantiate Patient Objects from Data-Set and Print them

```python
#creates a linear regression object
model = LinearRegression()

#creates patient objects for all rows of data in the csv file
Patient.instantiate_from_csv("main/patientdata.csv")

# sorts the data by age and then by whether they have dementia or not, thus two groups in numerical order
# and then prints all patients in all_patients
Patient.all_patients.sort(key = Patient.get_age, reverse = False)
Patient.all_patients.sort(key = Patient.get_dementia, reverse = True)
for patient in Patient.all_patients: 
    print(patient)
```

### Analysis 1: Brain pH of Dementia Patients Versus Non-Dementia Patients

### 1a. Create Lists and Variables

```python
# lists to hold all dementia patient brain pH values and no dementia brain pH values
dementia_ph = []
no_dementia_ph = []

# variables to hold the sums of all pH values for dementia and no dementia patients
dt_ph = 0.0
ndt_ph = 0.0

# variables to hold the mean of the pH values for dementia and no dementia patients to then use as the bars on the graph
d_bar = 0.0
nd_bar = 0.0

# variables to hold the standard deviations of the dementia and no dementia brain pH data sets
d_std = 0.0
nd_std = 0.0
```

### 1b. Retrieving pH from Patient Objects

```python
# the next two sections filter the data to separate the patients with dementia and then extract their brain pH into two separate lists
Patient.some_patients = Patient.filter_attr(Patient.all_patients, dementia = "Dementia")
for patient in Patient.some_patients:
    dementia_ph.append(patient.get_brain_ph())

Patient.some_patients = Patient.filter_attr(Patient.all_patients, dementia = "No dementia")
for patient in Patient.some_patients:
    no_dementia_ph.append(patient.get_brain_ph())

# summing of all pH values of each group to use to find the mean
for each in dementia_ph:
    dt_ph += each

for each in no_dementia_ph:
    ndt_ph += each
```

### 1c. Calculating Statistical Information

```python
# calculating the mean pH to use as the bar in the bar graph
d_bar = math.trunc(dt_ph * 1000 / (len(dementia_ph))) / 1000
nd_bar = math.trunc(ndt_ph * 1000 / (len(no_dementia_ph))) / 1000

# calculating the standard deviation of the two brain pH data sets
d_std = math.trunc(statistics.stdev(dementia_ph) * 1000) / 1000
nd_std = math.trunc(statistics.stdev(no_dementia_ph) * 1000) / 1000

# prints the value of the mean pH and standard deviations of the dementia and no dementia patient data sets
print("Statistical Information of Data Set:")
print(f'd_bar = {d_bar} | nd_bar = {nd_bar}') 
print(f'd_std = {d_std} | nd_std = {nd_std}')

# running the t-test of the data
t_stat, p_val = stats.ttest_ind(dementia_ph, no_dementia_ph)
t_stat = math.trunc(t_stat * 100) / 100
p_val = math.trunc(p_val * 100) / 100
print(f't_stat = {t_stat}, p_val = {p_val}')
print(f'')
```

### 1d. Creating the Bar Graph

```python
# creating lists for making the bar graph
average_ph_cols = ['Dementia', 'No Dementia']
mean_ph = [d_bar, nd_bar]
std_groups = [d_std, nd_std]
yerr = std_groups

# creates the limit of the y-axis based on the maximum pH value
ylim = max(mean_ph) + 1.5

# creating the bar graph
plt.bar(average_ph_cols, mean_ph, yerr=yerr, capsize=10, color=["orange", "blue"])
plt.title("Average Brain pH of Dementia Patients and Patients without Dementia")
plt.xlabel("Cognitive Status")
plt.ylabel("Average Brain pH")
plt.ylim(0, ylim)
plt.text(
        0.5, ylim - 1,
        f"t = {t_stat},\n p = {p_val}",
        ha = 'center',
        va = 'bottom'
        )

# saves the bar graph as a .png
plt.savefig("brain_ph_bar_graph.png")

# outputs the resulting bar graph
plt.show()
```

### Analysis 4: ABeta Ratio Versus Activated TAU Ratio

### 4a. Creating Lists and Filtering Patients into Groups

```python
# creating lists for data points for the scatterplot
dementia_ab = []
dementia_tau = []
no_dementia_ab = []
no_dementia_tau = []

# creates two lists that holds patients with either dementia or no dementia
dementia_patients = Patient.filter_attr(Patient.all_patients, dementia = "Dementia")
no_dementia_patients = Patient.filter_attr(Patient.all_patients, dementia = "No dementia")
```

### 4b. Retrieving Data From Patient Objects

```python
# the next two sections take the values for ab40/42 and t/ptau and calculates the ration between 
# the same type of protein and adding them into lists to use as x and y values in the scatterplot
for patient in dementia_patients:
    dementia_ab.append(patient.get_ab_ratio())
    dementia_tau.append(patient.get_tau_ratio())

for patient in no_dementia_patients:
    no_dementia_ab.append(patient.get_ab_ratio())
    no_dementia_tau.append(patient.get_tau_ratio())
```

### 4c. Creating the Bar Graph

```python
# creating x and y values for both patients with and without dementia
a = dementia_ab
b = dementia_tau
x = no_dementia_ab
y = no_dementia_tau

# determines the upper limits for the x- and y-axis based on the max value of the combined data set of protein ratios
if max(a) > max(x):
    xlim = max(x) + .05
else:
    xlim = max(a) + .05

if max(b) > max(y):
    ylim = max(b) + .05
else:
    ylim = max(y) + .05

# creation of the scatterplot
# red spots being patients with dementia and blue spots for patients without dementia
plt.scatter(a, b, s = 5, color='red')
plt.scatter(x, y, s = 5, color='blue')
plt.xlabel('Fraction of ABeta42 to Total ABeta')
plt.ylabel('Fraction of pTAU to tTAU')
plt.title('Scatter Plot Fraction of ABeta versus TAU')
plt.xlim(0, xlim)
plt.ylim(0, ylim)
```

### 4d. Linear Regression and Outputting the Scatterplot

```python
# reshapes the arrays to allow for linear regression to be computed
a_reg = np.array(a).reshape(-1, 1)
x_reg = np.array(x).reshape(-1, 1)

# linear regression of the data points of patients with dementia
model.fit(a_reg, b)
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(a_reg, b)
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR^2 = {r2:.2f}"
plt.text(xlim * .1, ylim * 0.95, equation, color = "red", fontsize = 12, verticalalignment = "top")
plt.plot(a_reg, model.predict(a_reg), color = "red")

# linear regression of the data points of patients without dementia
model.fit(x_reg, y)
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(x_reg, y)
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR^2 = {r2:.2f}"
plt.text(xlim * .6, ylim * 0.95, equation, color = "blue", fontsize = 12, verticalalignment = "top")
plt.plot(x_reg, model.predict(x_reg), color = "blue")

# outputs the resulting scatterplot
plt.show()

# saves the scatterplot as a .png
plt.savefig("abeta_versus_tau_ratios_scatterplot.png")
```

### Analysis 5: pTAU Concentration Versus MMSE Scores

### 5a. Creating Lists and Retrieving Patient Data

```python
# creates empty lists for pTAU and MMSE
ptau_mmse = []
mmse_scores = []

# add pTAU and MMSE values for patients who have an MMSE score
for patient in Patient.all_patients:
    if patient.mmse is not None:
        ptau_mmse.append(patient.ptau)
        mmse_scores.append(patient.mmse)

# prints the pTAU values and MMSE scores
print("pTAU values for MMSE analysis:")
print(ptau_mmse)

print("MMSE scores:")
print(mmse_scores)
```

### 5b. Linear Regression

```python
# Independent variable = pTAU
X = np.array(ptau_mmse).reshape(-1, 1)

# Dependent variable = MMSE score
y = np.array(mmse_scores)

# Do the linear regression
model.fit(X, y)

# Find slope, intercept, and R^2
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X, y)

# prints the slope, intercept, and R^2 values
print(f"Slope = {slope}")
print(f"Intercept = {intercept}")
print(f"R-squared = {r2}")

# combines the above characteristics of the line of best fit
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR² = {r2:.2f}"
```

### 5c. Creating the Scatterplot

```python
# creates a scatterplot of ptau levels vs mmse scores
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel("pTAU Concentration (pg/ug)")
plt.ylabel("MMSE Score")
plt.title("pTAU Concentration vs. MMSE Score")

# Put equation and R^2 on graph
plt.text(
    X.max(),
    y.max(),
    equation,
    fontsize=12,
    verticalalignment='top',
    horizontalalignment='right'
)

# saves the scatterplot as a .png file
plt.savefig("ptau_vs_mmse_regression.png")

# outputs the resulting scatterplot
plt.show()
```
